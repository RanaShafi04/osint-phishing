```python
df = pd.read_csv("../dataset/official_Phishing_Email.csv")
df.head(10)
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
#Drop tha Na values
df = df.dropna()
print(df.isna().sum())
```

```python
X = df.iloc[:,1:3001]
y = df.iloc[:,-1].values
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size = 0.25, random_state=5)
```

```python
X_train['Email Text'] = X_train['Email Text'].fillna('')
X_test['Email Text'] = X_test['Email Text'].fillna('')
```

```python
print(X_test.columns)
```

```python
vectorizer = TfidfVectorizer()
X_train_vectorized = vectorizer.fit_transform(X_train['Email Text'])
X_test_vectorized = vectorizer.transform(X_test['Email Text'])
```

# **Random Forest Classifier**

```python
rfc = RandomForestClassifier(random_state=4)

rfc.fit(X_train_vectorized,y_train)

print("Train Results \n")
y_train_pred  = rfc.predict(X_train_vectorized)
y_train_prob = rfc.predict_proba(X_train_vectorized)[:,1]

print("Confusion Matrix for Train : \n", confusion_matrix(y_train, y_train_pred))
print("Accuracy Score for Train : ", accuracy_score(y_train, y_train_pred))
print("ROC AUC for Train : ", roc_auc_score(y_train, y_train_prob))

print("+"*50)
print("Test Results \n")
y_test_pred  = rfc.predict(X_test_vectorized)
y_test_prob = rfc.predict_proba(X_test_vectorized)[:,1]

print("Confusion Matrix for Test : \n", confusion_matrix(y_test, y_test_pred))
print("Accuracy Score for Test : ", accuracy_score(y_test, y_test_pred))
print("ROC AUC for Test : ", roc_auc_score(y_test, y_test_prob))
```
